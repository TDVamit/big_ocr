```python
from pydantic import BaseModel, Field
from typing import List, Optional, Literal


class GeneralPolicyInfo(BaseModel):
    """general policy information and metadata"""

    date_of_application: str = Field(description="date the application was completed")
    carrier_name: str = Field(description="insurance carrier name")
    naic_code: str = Field(description="naic number of carrier")
    policy_number: Optional[str] = Field(None, description="policy number if renewing")
    effective_date: str = Field(description="start date of coverage")
    expiration_date: str = Field(description="end date of coverage")
    agency_name: str = Field(description="name of agency submitting application")
    agency_code: str = Field(description="assigned agency code")
    producer_name: str = Field(description="name of insurance producer")
    billing_plan: Literal["annual", "quarterly", "monthly"] = Field(
        description="billing frequency"
    )
    payment_method: Literal["eft", "credit_card", "check"] = Field(
        description="payment method"
    )
    quote_number: str = Field(description="quote reference number")


class NamedInsured(BaseModel):
    """named insured person details"""

    role: Literal["primary", "co_insured"] = Field(
        description="primary or co-insured role"
    )
    first_name: str = Field(description="first name of insured")
    last_name: str = Field(description="last name of insured")
    date_of_birth: str = Field(description="date of birth")
    phone_number: str = Field(description="contact phone number")
    email_address: str = Field(description="email address")


class Address(BaseModel):
    """address information"""

    street: str = Field(description="street address")
    city: str = Field(description="city")
    state: str = Field(description="state")
    zip_code: str = Field(description="zip code")


class Addresses(BaseModel):
    """residence and mailing addresses"""

    residence_address: Address = Field(description="primary residence address")
    mailing_address: Optional[Address] = Field(
        None, description="mailing address if different"
    )


class Vehicle(BaseModel):
    """vehicle details and specifications"""

    year: int = Field(description="model year")
    make: str = Field(description="manufacturer")
    model: str = Field(description="model name")
    body_type: str = Field(description="body type")
    vin: str = Field(description="vehicle identification number")
    use: Literal["pleasure", "commute", "business"] = Field(
        description="primary vehicle use"
    )
    garaging_address: str = Field(description="overnight parking location")
    annual_mileage: int = Field(description="estimated annual mileage")
    cost_new: float = Field(description="original purchase price when new")
    special_equipment_yn: bool = Field(description="has custom or special equipment")
    equipment_description: Optional[str] = Field(
        None, description="special equipment details"
    )
    anti_theft_device_yn: bool = Field(description="equipped with anti-theft device")
    passive_restraints_yn: bool = Field(
        description="equipped with airbags or passive restraints"
    )
    abs_yn: bool = Field(description="has anti-lock braking system")
    ownership: Literal["owned", "financed", "leased"] = Field(
        description="ownership type"
    )
    lease_term: Optional[int] = Field(
        None, description="lease term in months if leased"
    )
    lienholder_name: Optional[str] = Field(
        None, description="lienholder name if financed"
    )
    lienholder_address: Optional[str] = Field(None, description="lienholder address")


class Driver(BaseModel):
    """driver information and qualifications"""

    first_name: str = Field(description="driver first name")
    last_name: str = Field(description="driver last name")
    date_of_birth: str = Field(description="date of birth")
    gender: Literal["male", "female"] = Field(description="gender")
    marital_status: Literal["single", "married", "divorced", "widowed"] = Field(
        description="marital status"
    )
    license_number: str = Field(description="driver license number")
    license_state: str = Field(description="license issuing state")
    date_licensed: str = Field(description="date first licensed")
    relationship_to_applicant: str = Field(
        description="relationship to primary applicant"
    )
    occupation: str = Field(description="occupation")
    employer: str = Field(description="employer name")
    driver_training_yn: bool = Field(description="completed driver training course")
    good_student_yn: bool = Field(description="qualifies for good student discount")
    school_distance_gt_100mi_yn: bool = Field(
        description="student attends school over 100 miles away"
    )


class Coverage(BaseModel):
    """insurance coverage details and limits"""

    applies_to_vehicle_vin: str = Field(
        description="vin of vehicle or 'all' for policy-wide coverage"
    )
    liability_limit: str = Field(description="bodily injury and property damage limits")
    uninsured_motorist_bodily_injury: str = Field(
        description="uninsured motorist coverage limits"
    )
    underinsured_motorist_bodily_injury: str = Field(
        description="underinsured motorist limits"
    )
    uninsured_motorist_property_damage: Optional[str] = Field(
        None, description="uninsured motorist property damage limit"
    )
    personal_injury_protection_limit: Optional[str] = Field(
        None, description="personal injury protection amount"
    )
    comprehensive_deductible: Optional[int] = Field(
        None, description="comprehensive coverage deductible"
    )
    collision_deductible: Optional[int] = Field(
        None, description="collision coverage deductible"
    )
    medical_payments_limit: Optional[str] = Field(
        None, description="medical payments coverage amount"
    )
    rental_reimbursement_limit: Optional[str] = Field(
        None, description="rental car reimbursement limit"
    )
    towing_limit: Optional[str] = Field(
        None, description="towing and labor coverage limit"
    )
    oem_parts_coverage: bool = Field(
        description="original equipment manufacturer parts endorsement"
    )
    rideshare_coverage: bool = Field(description="rideshare coverage included")
    new_car_protection: bool = Field(description="new car replacement coverage")
    gap_coverage: bool = Field(description="loan or lease gap coverage")


class PriorInsurance(BaseModel):
    """previous insurance history"""

    prior_carrier_name: str = Field(description="previous insurance carrier name")
    prior_policy_number: str = Field(description="previous policy number")
    years_continuously_insured: int = Field(
        description="years of continuous insurance coverage"
    )
    any_lapses_in_coverage_yn: bool = Field(
        description="any coverage lapses in history"
    )
    assigned_risk_pool_yn: bool = Field(
        description="previously placed in assigned risk pool"
    )


class LossAccidentHistory(BaseModel):
    """loss and accident history record"""

    driver_license_number: str = Field(description="license number of involved driver")
    event_type: Literal["accident", "violation"] = Field(description="type of incident")
    date: str = Field(description="date of incident")
    description: str = Field(description="detailed incident description")
    at_fault_yn: bool = Field(description="driver was at fault")
    injuries_yn: bool = Field(description="incident involved injuries")
    cost: float = Field(description="total claim or incident cost")


class GeneralQuestions(BaseModel):
    """supplemental application questions"""

    any_vehicle_not_owned_by_applicant_yn: bool = Field(
        description="any listed vehicle not owned by applicant"
    )
    vehicle_use_other_than_personal_yn: bool = Field(
        description="vehicle used for delivery or rideshare"
    )
    any_modifications_to_vehicle_yn: bool = Field(
        description="any vehicle modifications made"
    )
    any_existing_vehicle_damage_yn: bool = Field(
        description="any pre-existing vehicle damage"
    )
    any_vehicles_uninsured_in_household_yn: bool = Field(
        description="any uninsured vehicles in household"
    )
    prior_policy_cancelled_or_declined_yn: bool = Field(
        description="ever had insurance cancelled or declined"
    )
    bankruptcy_or_judgment_last_5_years_yn: bool = Field(
        description="filed bankruptcy or had judgments in last 5 years"
    )
    applicant_in_military_yn: bool = Field(
        description="applicant currently in military service"
    )


class SignaturesAttachments(BaseModel):
    """signatures and required attachments"""

    applicant_signature: str = Field(description="applicant signature")
    producer_signature: str = Field(description="insurance producer signature")
    date_signed: str = Field(description="application signing date")
    acord_99_attached: bool = Field(description="accident history supplement attached")
    acord_101_attached: bool = Field(description="additional remarks form attached")


class CommercialAuto(BaseModel):
    """complete commercial auto insurance application"""

    general_policy_info: GeneralPolicyInfo = Field(
        description="policy information and metadata"
    )
    named_insureds: List[NamedInsured] = Field(description="all named insured parties")
    addresses: Addresses = Field(description="residence and mailing addresses")
    vehicles: List[Vehicle] = Field(description="all insured vehicles")
    drivers: List[Driver] = Field(description="all listed drivers")
    coverages: List[Coverage] = Field(description="coverage selections and limits")
    prior_insurance: PriorInsurance = Field(description="previous insurance history")
    loss_accident_history: List[LossAccidentHistory] = Field(
        description="claims and violation history"
    )
    general_questions: GeneralQuestions = Field(
        description="supplemental application questions"
    )
    signatures_attachments: SignaturesAttachments = Field(
        description="signatures and required forms"
    )
```
