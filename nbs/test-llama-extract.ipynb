```python
from llama_cloud_services import LlamaExtract
from pydantic import BaseModel, Field
from typing import List
from datetime import date
```

```python
# bring in our LLAMA_CLOUD_API_KEY
from dotenv import load_dotenv
load_dotenv()
```

```python
# Initialize client
extractor = LlamaExtract()
```

```python
class Vehicle(BaseModel):
    """Data model for a single vehicle on the policy."""
    make: str = Field(description="The manufacturer of the vehicle, e.g., Ford")
    model: str = Field(description="The model of the vehicle, e.g., F-150")
    year: int = Field(description="The model year of the vehicle")
    vin: str = Field(description="The unique 17-character Vehicle Identification Number")

class Driver(BaseModel):
    """Data model for a single driver on the policy."""
    first_name: str = Field(description="The first name of the driver")
    last_name: str = Field(description="The last name of the driver")
    license_number: str = Field(description="The driver's license number")

class CommercialAutoPolicy(BaseModel):
    """The main data model for extracting Commercial Auto policy details."""
    policy_number: str = Field(description="The unique identifier for the insurance policy")
    carrier_name: str = Field(description="The name of the insurance carrier, e.g., Progressive, Geico")
    effective_date: date = Field(description="The date the policy coverage begins")
    vehicles: List[Vehicle] = Field(description="A list of all vehicles covered by the policy")
    drivers: List[Driver] = Field(description="A list of all drivers covered by the policy")
```

```python
# Create extraction agent
agent = extractor.create_agent(name="insurance-parser", data_schema=CommercialAutoPolicy)
```

```python
commercial_auto_schema = CommercialAutoPolicy
agent.data_schema = commercial_auto_schema
agent.save()
```
